```python
def palindromic_product(n): 

    #smallest and biggest numbers to check 
    lowest = 10*10
    highest = 99*99

    #start palindrome at 0
    palindrome = 0 

    #for each number in the range of lowest, highest, going up by 1
    for num in range (lowest, highest, 1):
        for loc in range (num, highest, 1):
            total = num * loc

            #if total is smaller then the palindrome currently being looked at, the code stops  
            if total <= palindrome:
                break

            # turns the total into a string 
            string = str(total)
            #::-1 checks to see if it is the same when completely reversed
            if string == string[::-1]:
                palindrome = total

    return palindrome
```

```python
def primes_list (n):

    #automatically return nothing if the number is 1 or smaller
    if n <=1 :
        return []

    #starts the list with 2 
    primes = [2]

    # range checks only the odd numbers, going up to the user input n 
    for num in range(3, n+1, 2):
        # flag prime_num set to true 
        prime_num = True
        # find the square root of the number to see the top number to go to
        top = num**.5

        #look at each number in the primes list 
        for loc in primes:
            # if the number is larger then the top (the square root), then stop
            if loc > top:
                break
            #check if the prime divides into your number that you are counting up by
            elif num % loc == 0:
                # if it divides perfectly, then the number is not a prime number 
                prime_num = False
                break

        # if prime number is true at the end, add the number to the primes list
        if prime_num:
            primes.append(num)

    return primes
        
```

```python
primes_list(19)
```
