```python
def bubble_sort(arr, asc=True):
    length = len(arr)
    if asc:
        for iteration in range(length-1):
            for loc in range(length-1):
                if arr[loc]>arr[loc+1]:
                    (arr[loc], arr[loc+1]) = (arr[loc+1], arr[loc])

    return arr
```

```python
bubble_sort([1, 5, 7,8 ,2,3,4,5,3,2])
```
